```python
import random
Guess_num = random.randint(1,100)

attempt = 0 

while True:
    try:
        guessing_num = int(input("Guess the number from 1 to 100 : "))
    except ValueError:
        print("Invalid input! Please enter a number.")
        continue

    if guessing_num < 1 or guessing_num > 100:
        print("Please guess a number within the range 1–100.")
        continue
        
    attempt += 1

    if guessing_num == Guess_num:
        print("******Congratulation******")
        print(f"You guess it right : {guessing_num}")
        print(f"Total number of guesses is {attempt}")
        break
        
    if guessing_num > Guess_num :
        print(f"Too high : {guessing_num}")
    elif guessing_num < Guess_num :
        print(f"Too low : {guessing_num}")
```
